```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
# 世界座標系
world_x = 0
world_y = 200
```

```python
rad = math.pi / 180 * 45

x_shift = 200
y_shift = 0
```

```python
Affine = np.array([
    [np.cos(rad), -np.sin(rad), x_shift],
    [np.sin(rad), np.cos(rad), y_shift],
    [0, 0, 1]
])
```

```python
local =  Affine @ np.array([world_x, world_y, 1])

print(f"World coordinates: ({world_x}, {world_y})")
print(f"Local coordinates: ({local[0]}, {local[1]})")
```

```python
inv_Affine = np.linalg.inv(Affine)
local_inv = inv_Affine @ np.array([local[0], local[1], 1])
print(f"Inverse Local coordinates: ({local_inv[0]}, {local_inv[1]})")
```
